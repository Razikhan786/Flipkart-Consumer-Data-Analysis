### Fill missing Age values

```python

df['City'] = df['City'].fillna(df['City'].mode()[0])

df['Age'] = df['Age'].fillna(df['Age'].median())

df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

df['Income_Group'] = df['Income_Group'].fillna(df['Income_Group'].mode()[0])

df['Category'] = df['Category'].fillna(df['Category'].mode()[0])

df['Unit_Price_INR'] = df['Unit_Price_INR'].fillna(df['Unit_Price_INR'].median())

df['Discount_Percent'] = df['Discount_Percent'].fillna(df['Discount_Percent'].median())

df['Payment_Method'] = df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])

df['Shipping_Mode'] = df['Shipping_Mode'].fillna(df['Shipping_Mode'].mode()[0])

df['Delivery_Days'] = df['Delivery_Days'].fillna(df['Delivery_Days'].median())

df['Order_Status'] = df['Order_Status'].fillna(df['Order_Status'].mode()[0])
```

### Correct invalid ages

```python
df.loc[(df['Age'] < 18) | (df['Age'] > 100), 'Age'] = np.nan

df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].describe()
```

```python
df['Rating'].value_counts().sort_index()
```

```python
df.loc[(df['Rating'] < 1) | (df['Rating'] > 5), 'Rating'] = np.nan
```

```python
df['Rating'].value_counts().sort_index()
```

```python
df['Quantity'].value_counts().sort_index()
```

```python
df.loc[(df['Quantity'] <= 0) | (df['Quantity'] > 20), 'Quantity'] = np.nan

df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median()).round().astype(int)
```

```python
df['Quantity'].value_counts().sort_index()
```

```python
df['Discount_Percent'].describe()
```

```python
df.loc[
    (df['Discount_Percent'] < 0) | (df['Discount_Percent'] > 100),
    'Discount_Percent'
] = np.nan

df['Discount_Percent'] = df['Discount_Percent'].fillna(
    df['Discount_Percent'].median()
)
```

```python
df['Discount_Percent'].describe()
```

```python
df['Delivery_Days'].describe()
```

```python
df.loc[
    (df['Delivery_Days'] < 1) | (df['Delivery_Days'] > 30),
    'Delivery_Days'
] = np.nan

df['Delivery_Days'] = df['Delivery_Days'].fillna(
    df['Delivery_Days'].median()
)
```

```python
columns = ['Age', 'Unit_Price_INR', 'Order_Value_INR',
           'Discount_Percent', 'Delivery_Days']

for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print("Column:", col)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
    print("--------------------------------")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv("flipkart_cleaned_data.csv")

# Display first five records
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
df.describe(include='all')
```

```python
# Count customers from each city

city_customers = df['City'].value_counts()

print(city_customers)
```

```python
# Visualize top 10 cities

plt.figure(figsize=(12, 5))

city_customers.head(10).plot(kind='bar')

plt.title("Top 10 Cities by Number of Orders")
plt.xlabel("City")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.show()
```

```python
# Count orders in each product category

category_orders = df['Category'].value_counts()

print(category_orders)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(x=category_orders.index, y=category_orders.values)

plt.title("Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.show()
```

```python
# Calculate total revenue generated by each category

category_revenue = df.groupby('Category')['Order_Value_INR'].sum()

print(category_revenue.sort_values(ascending=False))
```

```python
plt.figure(figsize=(12, 6))

category_revenue.sort_values(ascending=False).plot(kind='bar')

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue (INR)")
plt.xticks(rotation=45)

plt.show()
```

```python
# Calculate average order value for each category

category_avg = df.groupby('Category')['Order_Value_INR'].mean()

print(category_avg.sort_values(ascending=False))
```

```python
# Complete category analysis

category_analysis = df.groupby('Category').agg(
    Total_Orders=('Order_ID', 'count'),
    Total_Quantity=('Quantity', 'sum'),
    Total_Revenue=('Order_Value_INR', 'sum'),
    Average_Order_Value=('Order_Value_INR', 'mean')
)

category_analysis = category_analysis.sort_values(
    by='Total_Revenue',
    ascending=False
)

category_analysis
```

```python
payment_analysis = df['Payment_Method'].value_counts()

print(payment_analysis)
```

```python
payment_revenue = df.groupby('Payment_Method')['Order_Value_INR'].sum()

print(payment_revenue.sort_values(ascending=False))
```

```python
payment_avg = df.groupby('Payment_Method')['Order_Value_INR'].mean()

print(payment_avg.sort_values(ascending=False))
```

```python
payment_summary = df.groupby('Payment_Method').agg(
    Total_Orders=('Order_ID', 'count'),
    Total_Revenue=('Order_Value_INR', 'sum'),
    Average_Order_Value=('Order_Value_INR', 'mean')
)

payment_summary.sort_values(by='Total_Revenue', ascending=False)
```

```python
# Count orders based on their status

order_status = df['Order_Status'].value_counts()

print(order_status)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Order_Status')

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=30)

plt.show()
```

```python
# Calculate average delivery time

average_delivery = df['Delivery_Days'].mean()

print("Average Delivery Days:", round(average_delivery, 2))
```

```python
# Compare delivery time for different shipping modes

shipping_analysis = df.groupby('Shipping_Mode').agg(
    Total_Orders=('Order_ID', 'count'),
    Average_Delivery_Days=('Delivery_Days', 'mean')
)

print(shipping_analysis.sort_values(by='Average_Delivery_Days'))
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x='Shipping_Mode',
    y='Delivery_Days',
    errorbar=None
)

plt.title("Average Delivery Time by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Average Delivery Days")

plt.show()
```

```python
status_revenue = df.groupby('Order_Status')['Order_Value_INR'].sum()

print(status_revenue.sort_values(ascending=False))
```

```python
# Category-wise revenue visualization

category_revenue = df.groupby('Category')['Order_Value_INR'].sum()

plt.figure(figsize=(10, 6))

category_revenue.sort_values().plot(
    kind='barh',
    figsize=(10, 6)
)

plt.title("Revenue by Product Category")
plt.xlabel("Total Revenue (INR)")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()
```

```python
# Convert order date into datetime format

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')

# Group sales by month

monthly_sales = df.groupby(
    df['Order_Date'].dt.to_period('M')
)['Order_Value_INR'].sum()

# Visualize monthly sales

plt.figure(figsize=(14, 6))

monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue (INR)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Count orders by status

status_count = df['Order_Status'].value_counts()

# Create pie chart

plt.figure(figsize=(8, 8))

plt.pie(
    status_count,
    labels=status_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Order Status Distribution")

plt.show()
```
